```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
car_df = pd.read_csv("/content/car_age_price (1).csv")
```

```python
car_df.head()
```

```python
car_df.shape
```

```python
car_df.duplicated().sum()
```

```python
car_df = car_df.drop_duplicates()
```

```python
car_df.shape
```

```python
# check correlation

car_df.corr()
```

```python
# Feature target splitting

y =car_df['Price']
x =car_df.drop('Price',axis=1)
```

```python
y
```

```python
x
```

```python
# train test split

from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=42)
```

```python
# create linear regression model

from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
model = lr.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_test
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error,r2_score
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
```

```python
year_2022 = pd.DataFrame({'Year': [2022]})

price_2022_linear = model.predict(year_2022)

print("Predicted price of a 2022 model:")
print(price_2022_linear[0])
```

Lasso regession

```python
from sklearn.linear_model import Lasso

```

```python
lasso_model = Lasso(alpha=1.0)

lasso_model.fit(x_train, y_train)

print("Lasso Regression model trained successfully")
```

```python
print("Lasso coefficient:")
print(lasso_model.coef_)

print("\nLasso intercept:")
print(lasso_model.intercept_)
```

```python
y_pred_lasso = lasso_model.predict(x_test)

print("Lasso predicted prices:")
print(y_pred_lasso)
```

```python
price_2022_lasso = lasso_model.predict(year_2022)

print("Predicted price of 2022 model using Lasso:")
print(price_2022_lasso[0])
```
