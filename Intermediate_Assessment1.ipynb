Q1. Write a function that return the range (maximum and minimum) of a given list of number



```python
def maximum(l):
  print(min(l))
def minimum(l):
  print(max(l))
```

```python
list1=[2,2,77,22,8,44,6,1,9,89]
maximum(list1)
minimum(list1)
```

Q2. Write function that returns mean value of a list of numbers


```python
def avg(l):
  print(sum(l)/2)
```

```python
list1=[2,2,77,22,8,44,6,1,9,89]
avg(list1)
```

Q3. Write a function that returns the median value of a list of numbers


```python
def median(l):
  length=len(l)
  mid=length//2
  if length%2==0:
    return (l[mid-1]+l[mid+1])/2
  else:
    return l[mid]
```

```python
list1=[2,2,77,22,8,44,6,1,9,89]
list2=list1.sort()
med = median(list1)
med
```

Q5. Write function to compute a given percentile value (e.g. 90th percentile) of a
list of numbers

```python
import pandas as pd
def p90(l):
  return l.quantile(0.90)
```

```python
list1=pd.Series([2,2,77,22,8,44,6,1,9,89])
per_90=p90(list1)
per_90
```
