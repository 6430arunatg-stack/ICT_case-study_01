TASK 8

```python
#Merge all four datasets into one DataFrame.

# Use the common column (country)
#Choose an appropriate join method.
```

```python
merged = population.merge(df_gdp, on = 'country', how ='inner')\
.merge(net_users, on='country',how ='inner' )\
.merge(df_lit, on='country',how ='inner' )
```

```python
merged
```

TASK 9

```python
#Create the following calculated columns:
#Internet Penetration (%)
# GDP Per Capita
```

```python
merged["internet_penetration (%)"] = (
    merged["internet_users"] / merged["population"] * 100
).round(2)
```

```python
merged["GDP_Per_Capita"] = (
    merged["GDP"] * 1000000 / merged["population"]
).round(2)
```

```python
merged
```

```python
#Create a SQLite database. Store the final merged dataset into a table named ‘country_statistics’
```

```python
import sqlite3
```

```python
conn =sqlite3.connect("country_statistics.db")
merged.to_sql("country_statistics",conn,
              if_exists = "replace",
              index = False
              )
conn.close()
```

Data Analysis Using SQL

```python
# Display the Top 5 countries by GDP. using sqlite
conn=sqlite3.connect('country_statistics.db')

pd.read_sql("""
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5
""", conn)
```

```python
#  Find the average literacy rate.

pd.read_sql("""
SELECT AVG(literacy_rate) AS
average_litracy_rate
from country_statistics
""", conn)
```

```python
# Display all countries having literacy rate above 90%.

pd.read_sql(
    """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
ORDER BY literacy_rate DESC
""",
    conn,
)
```

```python
# Find the Top 5 countries with the highest internet penetration.

pd.read_sql(
    """
SELECT country, `internet_penetration (%)`
FROM country_statistics
ORDER BY `internet_penetration (%)` DESC
LIMIT 5
""",
    conn,
)
```

```python
#Calculate the total population.

pd.read_sql(
    """
SELECT SUM(population) AS total_population
FROM country_statistics
""",
    conn,
)
```

```python
# Find the country with the highest GDP per capita.
pd.read_sql(
    """
SELECT country, GDP_Per_Capita
FROM country_statistics
ORDER BY GDP_Per_Capita DESC
LIMIT 1
""",
    conn,
)
```

```python
#Export the final merged dataset as a CSV file
df = pd.read_sql("SELECT * FROM country_statistics", conn)

df.to_csv("final_merged_dataset.csv", index=False)

print("CSV file exported successfully!")
```

```python
# Sort the dataset in descending order of GDP per Capita and display the Top 10 countries.
pd.read_sql( """
SELECT country,  GDP_Per_Capita
FROM country_statistics
ORDER BY  GDP_Per_Capita DESC
LIMIT 10
""", conn)
```
