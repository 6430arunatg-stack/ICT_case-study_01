```python
# Bank Account manager
def deposit(balance):
  amount = int(input("ENTER THE AMOUNT TO DEPOSIT : "))
  return balance + amount
def withdrawl(balance):
    withdrawl_amount =int(input("enter your withdrawl amount :"))
    if withdrawl_amount <= balance:
        return balance - withdrawl_amount
    else:
        print("Insufficient balance")
        return balance
def show_balance(balance):
    print("Balance:", balance)
```

```python
password=input('Enter password')
dig=0
upper=0
symb=0
symbol='!@#$%'
if len(password)>=8:
  for i in password:
    if i.isdigit():
      dig=dig+1
    if i.isupper():
      upper=upper+1
    if i in symbol:
      symb=symb+1
  if dig>0 and upper>0 and symb>0:
    print('Strong Password')
  else:
    print('Weak password!')
else:
  print('weak password!!')
```

```python

#fizzbuss variant
#Print numbers 1 to 50, but replace multiples of 3 with "Fizz",
#multiples of 5 with "Buzz",
#multiples of both with "FizzBuzz" — and multiples of 7 with "Bang".
for num in range(1,51):
  if num%3==0 and num<=50 and num>=1:
    print("Fizz")
  elif num%5==0 and num<=50 and num>=1:
    print("Buzz")
  elif num%3==0 and num%3==0 and num<=50 and num>=1:
     print("FizzBuss")
  else:
    print(num)
```
